```python
# dfs
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [], 'E': [], 'F': []
}
#정의 되었다고 생각을 하자 그러면 A가 방문하게되는 노드들은? 뭐얌?
visited=[]
def dfs(node):
    if node not in visited:
        visited.append(node)
        for next in graph[node]:
            dfs(next)
dfs('A')
print(visited)
```

```python
# bfs
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [], 'E': [], 'F': []
}
visited = [] # 방문한 곳 기록


from collections import deque

def bfs(startnode):
    visited=[startnode]
    queue=deque([startnode])
    while queue:
        a=queue.popleft()
        for i in graph[a]:
            if i not in visited:
                visited.append(i)
                queue.append(i)
    print(visited)
bfs('A')

    
```

```python
b=[]
def fibo(n):
    b=[0]*(n+1)# +1이유는 인덱스 0 일때 냅두려고
    b[1]=1
    b[2]=1
    for i in range(3,n+1):
        b[i]=b[i-1]+b[i-2]
    print(b)

fibo(10)
print(b)

# 내생각에는 이거서서 누적합 구할수 있을듯요
```
